2. Explain the difference between mutable and immutable data types with examples.
   - Mutable data types are those whose values can be changed after the object is created. This means you can add, remove, or modify elements without creating a new object. Common mutable types in Python include lists, dictionaries, and sets. For example, consider a list:
        

        fruits = ["apple", "banana", "mango"]
        fruits.append("orange")
        print(fruits)  # Output: ['apple', 'banana', 'mango', 'orange']


 In this case, the original list fruits is modified directly by adding a new item. The memory address of the list remains the same, meaning the object was changed in-place.

   - Immutable data types, on the other hand, cannot be changed once they are created. If you try to modify them, a new object is created instead. Examples include integers, floats, strings, and tuples. Here's an example using a string:

       
        greeting = "Hello"
        greeting += " World"
        print(greeting)  # Output: "Hello World"   
     
  Even though it looks like we're modifying the string, Python actually creates a new string object and assigns it back to the variable greeting. The original "Hello" string remains unchanged in memory.

  Understanding the difference between mutable and immutable data types is important because it affects how variables behave when passed to functions, stored in collections, or used in loops and assignments. Mutable types allow in-place changes, while immutable types provide safety from unintended side-effects.

4. Describe how dictionaries store data.
   - In Python, dictionaries store data in the form of key-value pairs. Each item in a dictionary has a unique key and an associated value. This structure allows for fast data retrieval because, unlike lists or tuples that use numeric indexes, dictionaries use keys to access values directly.

     Internally, Python uses a hash table to store dictionaries. When you create a key, Python runs it through a hash function, which generates a unique hash code. This hash code is used to determine where to place the key-value pair in memory. When you later access a value using the key, Python hashes the key again and jumps directly to the correct location, making lookups very fast.

     For example, in the dictionary student = {"name": "Bro", "age": 21}, "name" and "age" are keys, and "Bro" and 21 are their corresponding values. You can quickly access student["age"] to get 21 without searching through the entire dictionary.

      This key-value system is especially useful when you need to label data clearly and access it efficiently, such as storing user profiles, settings, or any structured data.

8. What is a hash table, and how does it relate to dictionaries in Python?
   - A hash table is an efficient data structure used to store key-value pairs, allowing quick access to values based on their keys. It works by using a hash function to convert a key (like a string or number) into a unique hash code, which is then mapped to an index in an underlying array. This index determines where the key-value pair is stored in memory.

     In Python, dictionaries are implemented using hash tables. When you create a dictionary like student = {"name": "Bro", "age": 21}, Python uses a hash function to process the keys "name" and "age" into hash codes. These hash codes point to specific memory locations where their corresponding values ("Bro" and 21) are stored. When you access student["age"], Python hashes the key "age" again, finds the memory index, and retrieves the value quickly—without having to search the entire dictionary.
     
     Hash tables offer very fast average time complexity for operations like:

        - Insertion: Adding a new key-value pair
        - Lookup: Accessing a value by key
        - Deletion: Removing a key-value pair

     This speed is typically O(1) (constant time), meaning it doesn't slow down as the size of the dictionary grows. However, if two keys hash to the same index (called a collision), Python handles this internally using techniques like open addressing or chaining, ensuring reliable performance.

     Because of their structure, dictionaries allow:

        - Fast access and updates
        - Flexible key types (as long as the key is immutable and hashable)
        - Efficient storage for large sets of data

     In summary, a hash table is the core mechanism behind Python dictionaries. It provides the ability to quickly map keys to values, which makes dictionaries one of the most powerful and frequently used data structures in Python programming.

14. How does the “in” keyword work differently for lists and dictionaries?
    - The in keyword is used to check for the presence of an item, but it behaves differently for lists and dictionaries in Python. It works differently for lists and dictionaries in Python because of how they are structured:
      
- For Lists:
  The in keyword checks if a value exists in the list.
  It searches through the list to find the value and returns True if it’s found, otherwise False.

              my_list = [1, 2, 3]
              print(2 in my_list)  # True (checks if 2 is in the list)
              print(4 in my_list)  # False (4 is not in the list)


- For Dictionaries:
  The in keyword checks if a key exists in the dictionary, not the value.
  It searches through the dictionary’s keys and returns True if the key exists, otherwise False.

              my_dict = {'a': 1, 'b': 2}
              print('a' in my_dict)  # True (checks if 'a' is a key in the dictionary)
              print(1 in my_dict)    # False (1 is not a key)

17. Describe the time complexity of accessing elements in a dictionary.
    - Accessing elements in a dictionary in Python typically has O(1) time complexity, meaning it's a constant-time operation. This efficiency comes from the underlying hash table data structure that dictionaries use.

     How It Works:
      - Hashing: When you use a key to access a value (e.g., my_dict[key]), Python applies a hash function to the key to quickly determine the index in the dictionary's internal structure.
      - Direct Access: The hashed index allows Python to jump directly to the location where the value is stored, resulting in fast access.

     Best Case (O(1)):
      - Most dictionary lookups take constant time—regardless of the size of the dictionary—because Python can access the key directly using the hash.

     Worst Case (O(n)):
      - In rare cases, such as when many keys hash to the same value (called hash collisions), the dictionary may need to check each key in the same "bucket" or index. This can degrade performance to O(n), where n is the number of elements in the dictionary.

     However, Python's implementation handles collisions efficiently, so this worst-case scenario is rare.

     Example:
             my_dict = {'a': 10, 'b': 20, 'c': 30}
             print(my_dict['b'])  # O(1) - Fast access to the value 20

19. Why are dictionaries considered unordered, and how does that affect data retrieval?
    - Dictionaries in Python were traditionally considered unordered because they did not maintain the insertion order of key-value pairs before Python 3.7. This meant that when you iterated over a dictionary, the order in which you added items was not guaranteed to be preserved.

     Behavior Before Python 3.7:
     - Dictionaries stored keys based on their hash values, not their insertion order.

     - As a result, the output of loops or printed dictionaries could vary.

     Behavior Since Python 3.7+:
     - Dictionaries now preserve insertion order by default.

     - However, they are still logically treated as unordered collections because the order is not used for accessing elements.

     How This Affects Data Retrieval:
      - Key-based access is not affected by order. You retrieve values using keys directly (dict[key]), which is always fast and consistent.

      - Order-sensitive operations like indexing (dict[0]) are not possible, unlike lists.

      - If your logic depends on the order of items, it's better to use OrderedDict (for older versions) or be cautious when using dict.
      
      Example:
          my_dict = {'a': 1, 'b': 2, 'c': 3}
          print(my_dict['b'])  # Efficient O(1) retrieval by key, not affected by order

20. Explain the difference between a list and a dictionary in terms of data retrieval.
    - The key difference between a list and a dictionary in terms of data retrieval is how you access elements.
     
     List: Index-Based Retrieval
       - Lists store elements in a sequential order.

       - You access data by position (index), which is always an integer starting from 0.

     Dictionary: Key-Based Retrieval
       - Dictionaries store data as key-value pairs.

       - The keys must be unique and hashable (like strings or numbers).
